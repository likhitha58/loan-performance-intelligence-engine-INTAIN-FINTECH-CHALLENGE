```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_PATH = Path("../data/raw/sf-loan-performance-data-sample.csv")

print("File exists:", DATA_PATH.exists())
print("File:", DATA_PATH)
```

```python
df_raw = pd.read_csv(
    DATA_PATH,
    sep="|",
    header=None,
    dtype=str
)

print("Shape:", df_raw.shape)
print("\nFirst 5 rows:")
display(df_raw.head())
```

```python
# Inspect how many non-null values exist in each raw column

non_null_counts = df_raw.notna().sum()

structure_check = pd.DataFrame({
    "raw_column": df_raw.columns,
    "non_null_count": non_null_counts.values,
    "null_count": df_raw.isna().sum().values,
    "sample_value": [
        df_raw[col].dropna().iloc[0] if df_raw[col].notna().any() else None
        for col in df_raw.columns
    ]
})

display(structure_check)
```

```python
# Identify completely empty columns

empty_columns = df_raw.columns[df_raw.isna().all()].tolist()

print("Completely empty columns:")
print(empty_columns)

print("\nNumber of completely empty columns:", len(empty_columns))
```

```python
# Check how many populated fields each row contains

fields_per_row = df_raw.notna().sum(axis=1)

print("Fields populated per row:")
print(fields_per_row.value_counts().sort_index())

print("\nMinimum populated fields:", fields_per_row.min())
print("Maximum populated fields:", fields_per_row.max())
```

```python
# Display the first record as raw positional fields

first_record = df_raw.iloc[0].to_frame(name="value")
first_record.index.name = "raw_column"

display(first_record)
```
